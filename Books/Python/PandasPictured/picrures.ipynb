### How to read this

Basically all we do is apply some operation to some object

An operation can result in different ways:
- Mutation - operation changes operand value
- Exposure - operation exposes some part of the object to us, if we change this, changes will happen in initial object
- Creation - operation results in something new, a copy or just an another value
- Side effect - nothing has changed, but operation did something (print function for example)

Sometimes they are combined, sometimes they have some nuances. This will be explicitly mentioned

Colors definitions

**Mutation**: operand and result have same type and denoted with different colors

**Exposure**: operand and result can have different type and denoted with same color.

In case they are same type, exposure will be noted explicitly

**Creation**: operand and result can have different types, denoted in different colors

First operand is always green. Something new is orange. Copies are blue.
