```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../datasets/diabetes.csv")
df.head()
```

```python
df.info()
```

```python
sns.boxplot(data=df, y="age", x="diabetes_risk")
plt.show()
```

```python
numeric_correlation = df.select_dtypes(include=["number"]).corr()

plt.figure(figsize=(10, 10))
sns.heatmap(data=numeric_correlation, annot=True, cmap="coolwarm", linewidths=1)
plt.show()
```
